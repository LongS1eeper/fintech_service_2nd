국민은행 앱 사용자 리뷰 데이터 JSON 파일을 읽어와서 csv 파일로 변환하기

```python
import json
import pandas as pd
```

```python
with open("./data/com_kbstar_kbbank.json", 'r', encoding = 'utf-8') as file:
    data = file.read()
```

```python
reviews = json.loads(data)
len(reviews)
```

```python
reviews[6]
```

```python
review_dict = {}

for dict in reviews:
    for key, value in dict.items():
        review_dict.setdefault(key,[]).append(value)
```

```python
review_dict_df = pd.DataFrame(review_dict)
review_dict_df.head()
```

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        if key == '별점':
            temp.append(int(review[key]))
        else:
            temp2 = str(review[key]).replace('!',' ').replace('.',' ').replace(',',' ')
            temp.append(temp2)
    result[key] = temp
len(result['별점'])
df = pd.DataFrame(result)
df
```

# 치킨집

```python
with open("./data/전국치킨집리스트.json", 'r', encoding = 'utf-8') as file:
    data2 = file.read()
```

```python
reviews2 = json.loads(data2)
len(reviews2)
```

```python
# 기본적으로 header와 body가 있음 그 중 정보는 body에 위치
reviews2[0]['body'].keys()
```

```python
# body는 dict 형식으로 그 중 items에 개별 가게 정보가 담겨있음
# items는 list 형식으로 모든 리스트가 동일한 키를 가짐
reviews2[0]['body']['items'][0].keys()
```

```python
reviews2[1]['body']['totalCount']
```

```python
chick_dict = {}

for list in reviews2:
    for list2 in list['body']['items']:
        for key, value in list2.items():
            chick_dict.setdefault(key,[]).append(value)
```

```python
reviews2[0]['body']['items'][0]['bizesNm']
```

```python
reviews2[0]['body']['items'][0]['ctprvnCd']
```

```python
chick_dict['bizesNm']
```

```python
chick_df = pd.DataFrame(chick_dict)
chick_df.head(5)
```

```python
len(chick_df)
```

```python
chick_dict_new = {}

for list in reviews2:
    for list2 in list['body']['items']:
        # zip 함수를 통해 묶음
        for key, value in zip(list['header']['columns'], list2.values()):
            chick_dict_new.setdefault(key,[]).append(value)
            
chick_df_new = pd.DataFrame(chick_dict_new)
chick_df_new.head(5)
```

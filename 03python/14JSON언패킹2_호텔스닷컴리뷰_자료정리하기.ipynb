```python
column = ['hotelId', '호텔이름', '호텔등급', '총리뷰수', '평점', '청결도', '서비스', '편안함','시설상태', '주변환경', '지역', '시도']
```

```python
hotel_dic = {}

for key in column:
    hotel_dic.setdefault(key,[])

for li in hotels:
    # hotelId
    if li['data']['body']['pdpHeader']['hotelId'] not in hotel_dic['hotelId']:

        hotel_dic['hotelId'].append(li['data']['body']['pdpHeader']['hotelId'])
        hotel_dic['호텔이름'].append(li['data']['body']['propertyDescription']['name'])
        hotel_dic['호텔등급'].append(li['data']['body']['propertyDescription']['starRatingTitle'])
        hotel_dic['총리뷰수'].append(li['data']['body']['reviewContent']['overall']['totalCount'])
        hotel_dic['평점'].append(li['data']['body']['reviewContent']['overall']['rating'])


        hotel_dic['청결도'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('cleanliness',None))
        hotel_dic['서비스'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('service',None))
        hotel_dic['편안함'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('comfort',None))
        hotel_dic['시설상태'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('condition',None))
        hotel_dic['주변환경'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('neighbourhood',None))


        hotel_dic['지역'].append(li['data']['body']['propertyDescription']['address'].get('region',None))
        hotel_dic['시도'].append(li['data']['body']['propertyDescription']['address']['locality'])
```

```python
hotel_dic = {}

for key in column:
    hotel_dic.setdefault(key,[])

for li in hotels:
    if li['data']['body']['pdpHeader']['hotelId'] is not in hotel_dic['hotelId']:
        hotel_dic['hotelId'].append(li['data']['body']['pdpHeader']['hotelId'])
        hotel_dic['호텔이름'].append(li['data']['body']['propertyDescription']['name'])
    hotel_dic['호텔등급'].append(li['data']['body']['propertyDescription']['starRatingTitle'])
    hotel_dic['총리뷰수'].append(li['data']['body']['reviewContent']['overall']['totalCount'])
    hotel_dic['평점'].append(li['data']['body']['reviewContent']['overall']['rating'])
    
    
    hotel_dic['청결도'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('cleanliness',"8.0"))
    hotel_dic['서비스'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('service',"7.6"))
    hotel_dic['편안함'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('comfort',"8.2"))
    hotel_dic['시설상태'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('condition',"7.8"))
    hotel_dic['주변환경'].append(li['data']['body']['reviewContent']['overall']['ratingAspects'].get('neighbourhood',"8.2"))
    
    
    hotel_dic['시도'].append(li['data']['body']['propertyDescription']['address']['locality'])
    
    if 'region' in li['data']['body']['propertyDescription']['address'].keys():
        hotel_dic['지역'].append(li['data']['body']['propertyDescription']['address']['region'])
    else:    
        gu = li['data']['body']['propertyDescription']['streetAddress'].split(" ")
        hotel_dic['지역'].append(gu[0])
 
```

```python
hotel_df = pd.DataFrame(hotel_dic)
hotel_df.head(5)
```

```python
review_dic = {}

column2 = ['hotelId', '리뷰일', '고객id', '여행타입', '평점', '리뷰']

for key in column2:
    review_dic.setdefault(key,[])

for li in hotels:
    for i in li['data']['body']['reviewContent']['reviews']['hermes']['groups']:
        for j in i['items']:
            review_dic['hotelId'].append(li['data']['body']['pdpHeader']['hotelId'])
            review_dic['리뷰일'].append(j['reviewDate'])
            review_dic['고객id'].append(j.get('itineraryId', None))
            review_dic['여행타입'].append(j['tripType'])
            review_dic['평점'].append(j['rating'])
            
            temp = str(j['description'])
            temp = temp.replace('\n',' ')
            
            review_dic['리뷰'].append(temp)
```

```python
review_df = pd.DataFrame(review_dic)
review_df.tail(3)
```

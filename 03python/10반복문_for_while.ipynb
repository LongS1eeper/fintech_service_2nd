# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있음
* for는 **반복 횟수가 정해져 있는 경우** 사용
* while은 반복 횟수가 정해져 있지 않은 경우 **(무한 반복)** 사용

## for 반복문: 반복할 횟수가 정해져 있는 경우
for 변수1, 변수2 in 반복할 객체/자료(문자열, 리스트, 튜플, 딕셔너리): <br>
____반복할 코드

```python
# 숫자 생성 함수 range(시작, 끝, step)
```

```python
# for 문을 이용하여 1-10 출력
for i in range(1,11):
    print(i, end=" ")
```

* 순서가 있는 자료형 (문자열, 리스트, 튜플) + 순서는 없지만 dict도 반복문에서 사용 가능

```python
# 문자열
for char in "파이썬은 문자열을 반복문에 넣어 출력 가능":
    print(char, end="@")
```

```python
# 리스트
for l_item in [1,2,3,"사", "오", [6,7,8],(8,9,10), {"key":"value"}]:
    print(l_item, end="@")
```

```python
d=dict(이름="홍길동", 나이=40, 주소="서울",전화번호="010-1324-5678")
```

```python
for item in d:
    print(item)     # key가 출력
    
for item in d:
    print(d.get(item,'key 없음'))  # value가 출력 (key가 없으면 'key 없음' 출력)
```

```python
for key, value in d.items():
    print(key, ':', value)
```

* 1부터 100까지의 수를 만들고 그 중에서 30-70까지마니 더해서 출력하세요    
* 과정 출력
* 2가지 방법을 이용해서 코드 작성 if 조건문, 슬라이싱

```python
hap = 0
for i in range(1,101):
    if 30 <= i <= 70:
        hap += i
    print(f"숫자 = {i} / 합 = {hap}")
```

* 100에서 1까지 더하는 반복문을 만드세요
* range에서 역순으로 숫자를 생성
* range(1,101)을 뒤집는 방법 2가지

```python
hap = 0
for i in range(100,0,-1):
    hap+=i
hap
```

```python
hap = 0
for i in range(1,101)[::-1]:
    hap += i
hap
```

```python
hap = 0
for i in reversed(range(1,101)):
    hap += i
    
hap
```

## enumerate() 반복문에서 자료의 인덱스를 만들어주는 함수

```python
s = "enumerate 함수는 반복문에서 index를 만들어 출력해주는 함수입니다."
```

```python
for idx, i in enumerate(s):
    print(idx, i)
```

## continue, break
* continue: 코드 실행을 건너뛰고 반복 계속 진행
* break: 특정 조건이 되었을 때 반복문 정지

```python
import time
for idx, i in enumerate(s):
    print(f"현재 반복 횟수 : {idx+1}/{len(s)}", end = '\r')
    if idx % 2 == 0:
        continue
    else:
        if idx > 19:
            break
            print(idx,i)
    time.sleep(1)
```

while 문으로 1-10 출력하기

```python
i = 1
while(i<11):
    print(i, end=" ")
    i +=1
```

20에서 1까지 숫자를 더한 값을 출력

```python
i = 20
hap = 0
while i>=1:
    hap += i
    i -= 1

print(hap)
```

* 1-50 까지 수 중에서 3의 배수이거나, 짝수인 수를 찾아서 더한 값을 출력하세요.

```python
i = 1
hap = 0
while(i<=50):
    if i % 3==0 or i % 2 == 0:
        hap += i
    i+=1
print(hap)
```

## while의 무한 반복
* whjle 뒤의 조건이 True 인 경우 무한 반복
* while문 내에서 if 조건문으로 break를 걸어주어야 멈춤

```python
# 무한 반복
while 1:
    print("무한반복")
    break
```

* 무한 반복이 되는 조건: True
* True가 되는 경우: 1, 정수, 실수, 문자열, list, tuple, dict, True
* False가 되는 경우: 0,0.0,"",'',[],(),{},False

# 중첩 반복문
* 반복문을 2개 이상 겹쳐서 사용

```python
for num1 in range(2,10):
    for num2 in range(1,10):
        print(f"{num1} X {num2} = {num1*num2}")
```

* 반복문 3개로 시계 만들기
* 초 00 - 59
* 분 00 - 59
* 시 00 - 23
* import time
* time.sleep(0.2)

```python
import time

for hour in range(0,24):
    for min in range(0,60):
        for sec in range(0,60):
            print(f"{hour:02d}:{min:02d}:{sec:02d}초", end="\r")
            time.sleep(0.5)
```

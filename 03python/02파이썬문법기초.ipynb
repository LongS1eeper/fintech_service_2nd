# print 문
* print 함수 안에있는 내용을 출력해주는 함수
* print(출력할 내용)
* print(1) 출력할 내용이 숫자인 경우, 변수인 경우 따음표 X
* print("hello world") 출력할 내용이 문자, 문자열인 경우 반드시 따옴표 필요

```python
print(1)
```

```python
# 따옴표까지 같이 출력하고 싶을때는 섞어서
print("이것은 '파이썬'이다.")
```

# print문의 옵션
* sep: 구분자 - print문 안에 여러개의 자료를 출력할 때 자료를 구분해주는 문자
* 기본값은 " " = 공백

```python
print("이것은", '파이썬', '입니다.')
```

```python
print("이것은", "파이썬",'입니다.', sep="_")
```

```python
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ")
```

* END옵션 - print가 끝나고 실행될 옵션
* 기본값 = \n - 이스케이프 문자, new line, 새줄

```python
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ", end="")
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ")
```

```python
# end 값을 "\t" (tap)으로 준 경우
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ", end="\t")
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ")
```

```python
# end 값을 공백 "\r" (carriager return)으로 준 경우
print("헐크", "아이언맨",'스파이더맨.', sep=" vs ", end="\r")
print('스파이더맨',"아이언맨", "헐크헐크", sep=" vs ")
```

```python
# end 값을 공백 "\n" (new line)으로 준 경우
print("헐크", 1, 3.14, sep="-", end="\n\n\n")
print('스파이더맨',"아이언맨", "헐크헐크", sep=" vs ")
```

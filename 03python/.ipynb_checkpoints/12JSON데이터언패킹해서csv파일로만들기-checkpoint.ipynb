국민은행 앱 사용자 리뷰 데이터 JSON 파일을 읽어와서 csv 파일로 변환하기

```python
import json
import pandas as pd
```

```python
with open("./data/com_kbstar_kbbank.json", 'r', encoding = 'utf-8') as file:
    data = file.read()
```

```python
reviews = json.loads(data)
len(reviews)
```

```python
reviews[6]
```

```python
review_dict = {}

for dict in reviews:
    for key, value in dict.items():
        review_dict.setdefault(key,[]).append(value)
```

```python
review_dict_df = pd.DataFrame(review_dict)
review_dict_df.head()
```

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        if key == '별점':
            temp.append(int(review[key]))
        else:
            temp2 = str(review[key]).replace('!',' ').replace('.',' ').replace(',',' ')
            temp.append(temp2)
    result[key] = temp
len(result['별점'])
df = pd.DataFrame(result)
df
```

# 치킨집

```python
with open("./data/전국치킨집리스트.json", 'r', encoding = 'utf-8') as file:
    data2 = file.read()
```

```python
reviews2 = json.loads(data2)
len(reviews2)
```

```python
reviews2[0]['body'].keys()
```

```python
reviews2[0]['body']['items'][0].keys()
```

```python
reviews2[1]['body']['totalCount']
```

```python
chick_dict = {}

for list in reviews2:
    for list2 in list['body']['items']:
        for key, value in list2.items():
            chick_dict.setdefault(key,[]).append(value)
```

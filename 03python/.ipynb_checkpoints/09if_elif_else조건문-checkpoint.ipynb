# if else 조건문

## if 조건문 : 실행 else : 실행
if 조건식 :

___ 조건이 참일 때 실행할 코드

else:

---- 조건이 거짓일 떄 실행할 코드

```python
l1 = ['파이썬', "조건문", 10, 15.3, True, [1,2,3], (4,5,6), {"key":"value"}]
```

```python
if "파이썬" in l1:
    print("l1안에 파이썬이 있습니다.")
else:
    print("l1안에 파이썬이 없습니다.")
```

* 숫자를 입력 받아 홀수인지 짝수인지 판단하는 프로그램

```python
a = int(input("숫자를 입력하세요 : "))

if a %2 == 0:
    print(f"{a}는 짝수입니다.")
else: 
    print(f"{a}는 홀수입니다.")
```

```python
x = int(input())
if 0 <= x <= 10 :
    print(f"{x}는 0에서 10사이에 있는 숫자입니다.")
else:
    print(f"{x}는 0에서 10사이에 있는 숫자입니다.")
```

## 2. 다중 조건문 사용하기 elif
if 조건식 : <br>
___ 조건이 참일 때 실행할 코드<br>
elif 조건식2 :<br>
___ 조건식2가 참일 때 실행할 코드<br>
elif 조건식 3:<br>
___ 조건식3이 참일 때 실행할 코드<br>
else:<br>
___ 조건식 1,2,3이 모두 거짓일 때 실행할 코드<br>

점수를 입력 받아서 81-100: A, 61-80: B, 41-60: C. 21-40: D, 0-20: F 등급

```python
score = int(input("점수를 입력하세요."))
if 0 <= score <= 100:
    if 0 <= score <= 20:
        print(f"{score}점은 F 등급입니다.")
    elif 21<= score <= 40:
        print(f"{score}점은 D 등급입니다.")
    elif 41 <= score <= 60:
        print(f"{score}점은 C 등급입니다.")
    elif 61 <= score <= 80:
        print(f"{score}점은 B 등급입니다.")
    else: 
        print(f"{score}점은 A 등급입니다.")
else:
    print("숫자를 0-100 사이로 입력해 주세요.")
```

* 1-50 사이의 숫자를 입력받아서 3의 배수면 3의 배수 입니다. 3의 배수가 아니면 3의 배수가 아닙니다.
    * 입력받기, 정수로 변환하기
    * 1-50 사이의 숫자인가 ?
    * 3의 배수인가 ?
    * 출력

```python
num = int(input("1-50 사이의 숫자를 입력하세요"))
if 1 <= num <= 50:
    if num % 3 == 0 :
        print(f"{num}은 3의 배수 입니다.")
    else:
        print(f"{num}은 3의 배수가 아닙니다.")
else:
    print("숫자를 1-50 사이로 입력해주세요.")
```

## 3. 조건 표현식
* 참일 때 실행조건 if 조건 else 거짓일 때 실행 코드

```python
# 기존 if else 문
num = int(input("숫자를 입력하세요"))
if num % 2 == 0:
    print("짝수")
else :
    print("홀수")
```

```python
# 조건 표현식
num = int(input("숫자를 입력하세요"))
print("짝수") if num % 2 == 0 else print("홀수")
```

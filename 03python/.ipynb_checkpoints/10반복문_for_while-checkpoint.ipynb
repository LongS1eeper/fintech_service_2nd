# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있음
* for는 **반복 횟수가 정해져 있는 경우** 사용
* while은 반복 횟수가 정해져 있지 않은 경우 **(무한 반복)** 사용

## for 반복문: 반복할 횟수가 정해져 있는 경우
for 변수1, 변수2 in 반복할 객체/자료(문자열, 리스트, 튜플, 딕셔너리): <br>
____반복할 코드

```python
# 숫자 생성 함수 range(시작, 끝, step)
```

```python
# for 문을 이용하여 1-10 출력
for i in range(1,11):
    print(i, end=" ")
```

* 순서가 있는 자료형 (문자열, 리스트, 튜플) + 순서는 없지만 dict도 반복문에서 사용 가능

```python
# 문자열
for char in "파이썬은 문자열을 반복문에 넣어 출력 가능":
    print(char, end="@")
```

```python
# 리스트
for l_item in [1,2,3,"사", "오", [6,7,8],(8,9,10), {"key":"value"}]:
    print(l_item, end="@")
```

```python
d=dict(이름="홍길동", 나이=40, 주소="서울",전화번호="010-1324-5678")
```

```python
for item in d:
    print(item)     # key가 출력
    
for item in d:
    print(d.get(item,'key 없음'))  # value가 출력 (key가 없으면 'key 없음' 출력)
```

```python
for key, value in d.items():
    print(key, ':', value)
```

* for 반복문을 이용해서 1부터 10까지 정수를 더한 합을 출력하세요

```python
sum=0
for i in range(1,11):
    sum += i
print(sum)
```

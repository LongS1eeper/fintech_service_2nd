# 리스트 안에서 원하는 값의 인덱스 찾기
* in, not in: 리스트 안에 원하는 값이 있는지 확인
* 원하는 값이 있는 위치를 찾을 때 .index(찾을 값
* 동일한 값이 여러개 있는 경우 가장 앞의 인덱스를 반환
* 2번쨰 이상의 자료를 찾으려면 옵션에 시작 인덱스를 지정하면 된다.
* .index(찾는 값, 시작 인덱스 번호)

```python
4 in a
```

```python
4 in a+b
```

```python
3 in [1,2,[3,4,]]      # 리스트 속 리스트의 요소는 찾을 수 없음
```

```python
lis = [1,2,[3,4,]]
3 in lis[2]
```

```python
a.index(3)
```

a에서 3번 인덱스부터 3을 찾으세요

```python
a=[1,2,3,4,5,6,7,8,9,4,3]
a.index(3,3)
```

# 튜플 tuple
* 튜플은 리스트처럼 모든 자료형을 넣어 만들 수 있다.
* 순서가 있는 자료형, 인덱스가 있다.
* 인덱싱, 슬라이싱 가능
* 자료의 추가 삭제가 안됨
* ()안에 자료를 넣어서 만든다.
* 튜플 형변환 함수는 tuple()
* 자료 1개만 넣어서 만들 수 있다. (1,) 형식이어야 한다.
* 튜플 생성시 자료가 2개 이상인 경우는 () 생략 가능

```python
t = (1,2,3)
```

```python
t[0]
```

```python
t[::-1]
```

## 튜플 + 튜플 = 두개의 튜플의 요소가 하나로 합쳐짐

```python
t1 = (1,2,3)
t2 = (4,5,6)
t1 + t2
```

* 리스트와 같이 특정 값의 인덱스를 찾는 것은 가능

```python
t.index(2)
```

* 튜플에 자료를 추가하고 싶을 땐 list로 변환
* 튜플에 자료를 1개만 넣고싶을 때 -> (자료,)
* 괄호를 생략하고 튜플만들기

```python
t5 ="튜플", "괄호없이?", "만들기"
type(t5)
```

# 시퀀스 객체(자료형)
* 문자열, 리스트, 튜플 같은 순서가 있는 자료형: 시퀀스 자료형
* 순서가 있고, 인덱스 번호가 부여되며 인덱싱, 슬라이싱이 가능
* 반복문에 시퀀스 자료형을 그대로 넣어서 반복시키는 것이 가능
* 시퀀스 객체는 + 연산자를 이용해서 서로 연결이 가능

```python
"시퀀스" + "자료형"
```

```python
[1,2,3,4] + [5,6,7,8]   # extend와 기능 동일
```

```python
(1,2,3,4) + (5,6,7,8)
```

시퀀스 객체는 *로 반복 가능

```python
[3,2]*3
```

* 리스트, 튜플은 sum(), max(), min()과 같은 함수로 연산도 가능

```python
t1
```

```python
sum(t1)
```

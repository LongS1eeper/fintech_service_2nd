# 딕셔너리 사용하기

## dict 함수를 사용해서 dict 만들기
* dict(key1=value1, key2=value2, key3=value3)
* key에 ' ', " "를 사용하지 않는다.
* key는 모두 문자형으로 반환된다.
* value가 문자형 자료면 반드시 ""가 필요하다.


```python
dict_ex1 = dict(정수=100, 실수=3.5, 문자="문자열", 
                리스트 = [1,2,3], 튜플 = (1,2,3), 딕셔너리 = {'key':'value'}
               )
dict_ex1
```

* dict([(key1, value1), (key2,value2)])
* 리스트 안에 key, value 쌍으로 된 튜플을 넣어서 만듦

```python
dict_ex2 = dict([('정수',100), ('실수',3.5), ('문자',"문자열"), 
                ('리스트', [1,2,3]), ('튜플', (1,2,3)), ('딕셔너리', {'key':'value'}
               )])
dict_ex2
```

* dict(zip([key1, key2, key3, key4], [value1, value2, value3, value4]))
* key 묶음 리스트와 value 묶음 리스트를 zip 함수로 합친 후 dict를 생성
* key 묶음의 요소 개수와 value 묶음의 요소 개수가 일치해야 함

```python
dict_ex3 = dict(zip())
```

```python
for a,b in zip([1,2,3,4],[5,6,7,8]):
    print(a,b)

for a in zip([1,2,3,4],[5,6,7,8]):
    print(a)
```

```python
dict_ex3 = dict(zip(['정수','실수','문자열','리스트','튜플','딕셔너리'],
                   [10,13.5,'문자열입니다.',[1,2,3],(1,2,3),{10:30}]))
dict_ex3
```

## 딕셔너리 함수 사용하기

### setdefault(key, value)
* 딕셔너리에 key와 value를 추가
* key가 중복되면 나중에 추가하는 값을 무시함
* key만 넣고 value를 생략하면 자동으로 None이 입력됨

```python
d
```

```python
d['년월']=20200412
```

```python
d
```

```python
d.setdefault('년월',20202020)
```

### update(key=value)
* 기존의 key가 있는 경우 값을 새로운 값으로 update
* 기존에 key가 없는 경우 새로운 key, value 생성
* key 부분은 '," 을 사용하지 않음

```python
d.update(방문자수=500)
d
```

```python
d.update(국가='영국')
d
```

## 딕셔너리에서 value를 가져오는 함수.get()
* .get(key, key가 없을 때 기본 값)
* 딕셔너리 내에 key가 있다가 없다가를 반복하는 경우 오류 방지

```python
# 강수량이라는 key가 있는 경우 value 값을, 없으면 default 값
d.get('강수량','default')
```

## 딕셔너리에서 key, value, (key,value)를 가져오는 함수
* keys(): 딕셔너리에서 key만 출력
* values(): 딕셔너리에서 value만 출력
* items(): 딕셔너리의 key 와 value를 튜플 묶음 형태로 출력

* 모두 list()로 형변환 해주면 리스트처럼 이용 가능
* 반복문에서 이용 가능

```python
d.keys()
```

```python
d_keys = list(d.keys())
d_keys[1]
```

```python
for i in d.keys():
    print(i)
```

```python
d.values()
```

```python
for j in d.values():
    print(j)
```

```python
d.items()
```

```python
list(d.items())
```

```python
for k,v in d.items():
    print(v)
```

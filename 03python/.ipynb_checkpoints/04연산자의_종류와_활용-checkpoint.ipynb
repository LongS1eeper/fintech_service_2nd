* 산술, 할당, 비교, 논리, 식별, 멤버

# 산술연산자 +, -, *, /, %, //, **

```python
1 + 1
```

```python
3 - 4
```

```python
5 * 3
```

```python
4 / 2   # 무조건 실수 도출
```

```python
7 % 3    # 나머지를 구하는 연산자 (홀수짝수 구분 용이)
```

```python
3 ** 4
```

```python
a = 10
b = 5
```

# 비교연산자
* == : 같다
* != : 다르다
* \> : 초과
* \< : 미만
* \>= : 이상
* <= : 이하

```python
a = 3
b = 3.0
c = 5
d = "Python"
e = "python"
f = True
g = False
```

* == 같으면 True, 다르면 False

```python
a == b
```

```python
d == e   # 대소문자 구분
```

* != 다르면 True, 같으면 False

```python
d != e
```

```python
a > b, c <= 7
```

# 논리 연산자
* and, or, not

* and는 두 값이 모두 참이어야 참

```python
True and True
```

```python
True and False
```

* or 두 값 중 하나라도 참이면 참, 두 값 모두 거짓이어야 거짓

```python
True or True
```

```python
True or False
```

```python
False or False
```

not 값을 반대로 뒤집음

```python
not False
```

* and, or, not의 우선순위
* not -> and -> or 순으로 판단

```python
not True and False or not False
```

# 식별연산자
* is, is not (object, data type)이 같은지 비교

```python
print(a)
print(b)
print(c)
```

```python
print(a==b)   # a는 정수, b는 실수 3.0이라 값이 같음
```

```python
print(a is b)    # a는 정수 3, b는 실수 3.0이라 데이터 타입이 달라 False
```

```python
d = 3
```

```python
print(a is d)    # a와 d가 모두 int 3이기 때문에 True
```

```python
print(a is not b)
```

# 멤버연산자
* in. not in

```python
mem = [1,2,3,4,5]
```

```python
3 in mem
```

```python
10 not in mem
```

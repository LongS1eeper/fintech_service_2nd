* 매개변수 3개짜리 함수
* num_make라는 함수를 매개변수 3개를 이용해서 만드세요.
* 숫자를 3개 입력받아 첫번째 숫자는 시작숫자, 두번째 숫자는 끝 숫자, 마지막 숫자는 간격
* 먼저 숫자를 생성한 결과를 출력하고
* 숫자 리스트에 *10한 결과를 다시 출력하세요

```python
def num_maker(a,b,c):
    result = list(range(a,b,c))
    result2 = []
    print(result)
    for i in result:
        result2.append(i*10)
    print(result2)
    return result, result2
```

```python
num_maker(1,11,2)
```

* num_maker2 = 정방향, 역방향 리스트가 출력되는 함수
* 3개의 숫자를 입력받아서 숫자 리스트를 만들고
* 리스트 +3을 한 결과를 리스트로 반환
* 한 번은 정방향으로 출력
* 또 한 번은 역방향으로 출력

```python
def num_maker2(start,end, step):
    list_result = list(range(start,end,step))
    reversed_list = list_result[::-1]
    return list_result, reversed_list
```

```python
result = num_maker2(1,30,2)
print(result)
```

# 중간에 함수를 종료시키기 위한 용도의 return
* 반복문에서 break처럼 사용 가능
* 함수 내에서 return이 실행되면 함수는 종료

* 1에서 100까지 숫자 중에서 짝수만 더해서 출력하시오.
* 단, 숫자가 50 이상이 되면 멈추시오.
* 더해지는 과정도 출력

```python
def even(num):
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 ==0:
                hap += i
                print(hap, end=" ")
        else:
            break
    return hap
```

```python
even(list(range(1,101)))
```

```python
def even2(num):
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 ==0:
                hap += i
                print(hap, end=" ")
        else:
            return hap
```

```python
even2(list(range(1,101)))
```

* 점수를 입력 받아서 90-100이면 A학점, 81-90이면 B학점, 71-80이면 C학점, 61-70이면 D학점, 60 이하는 E학점

```python
def student_grade(num):
    if 0 <= num <= 100:
        if num >= 90 :
            return 'A'
        elif num >= 81:
            return 'B'
        elif num >= 71:
            return 'C'
        elif num >= 61:
            return 'D'
        else:
            return 'F'
    else:
        return "점수를 다시 입력하세요."
```

```python
score = int(input("점수를 입력하세요: "))
grade = student_grade(score)
print('등급은',grade)
```

## 인수: 함수에 집어넣는 자료 (매개변수에 넣는 자료)
#### 위치인수(positional argument):
* 위치가 고정된 인수
* 순서가 바뀌니까 개수가 달라지면 오류
#### 가변위치인수(args,Variable-length Arguments)
* 여러개의 자료를 넣을 수 있는 인수, 제한 없음
#### 키워드 인수(keyword arguments)
* 매개변수에 이름을 명시해서 인수를 넣어주는 것
#### 가변 키워드 인수(kwrags, variable-length keyword arguments)
* 키워드 인수를 숫자 제한 없이 넣는 경우

* 위치인수(positional argument)

* my_sum
* 여러개의 숫자를 입력받으면 합계를 구해주는 함수

```python
def my_sum(*num):
    hap = 0
    for i in num:
        hap += i
    return hap
```

```python
result = my_sum(*list(range(1,100)))
print(result)
```

* 위치인수와 가변인수 동시에 받기
* 가변인수는 무조건 위치인수보다 뒤에 와야함

```python
# 정수와 리스트를 입력받아 리스트의 요소에 입력받은 정수를 곱하는 함수
```

```python
def muls(a,*args):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

* 키워드 가변인수<br>
def 함수명(**kwargs):

```python
def personal_info2(**kwargs):       # key:value 인식
    print(kwargs)
    key = list(kwargs.keys())
    value = list(kwargs.values())
    print(key)
    print(value)
```

```python
personal_info2(address= '서울시 강남구', name='홍길동')
```

* 위치인수와 키워드 가변 인수가 같이 사용하기
* 위치인수 다음에 키워드 가변인수가 와야한다.

```python
def personal_info3(num, **kwargs):
    print(kwargs)
    key = list(kwargs.keys())
    value = list(kwargs.values())
    print(num)
    print(key)
    print(value)
```

```python
personal_info3(1,address= '서울시 강남구', name='홍길동')
```

* 가변인수와 키워드 가변인수 함께 쓰기
* 가변인수 => 키워드 가변인수 순으로 쓴다
* def 함수명(*args, **kwargs)

```python
def custom_print(*args, **kwargs):
    print(args, kwargs)
    print(*args, **kwargs)     # *은 풀어낸다는 의미
```

```python
custom_print(1,2,3,sep="||", end="!!")
```

#### 매개변수 = 함수에서 자료를 받을 때 사용하는 변수
* 위치인수(a): 위치가 고정된 매개변수, 위치인수는 키워드 인수로 사용 가능(변수명 = 값)
* 가변인수(*args): 여러 개의 자료를 동시에 받을 수 있음 => tuple로 변환되어 들어옴
* 키워드 가변인수(**kwargs): 여러 개의 자료를 동시에 받을 수 있는데 key=value 형태로 받는 인수 => dict로 들어옴

* 매개변수에 초기값 지정하기
* 초기값이 지정된 매개변수는 반드시 마지막에 위치해야 함

* 위치인수, 키워드인수, 가변인수, 키워드 가변인수를 동시에 다 사용할 경우
* 위치인수 => 키워드 인수(기본값), 가변인수, 키워드 가변인수 순서로 사용한다.

```python
def personal_info6(idx, address="비공개", *args, **kwargs):
    print(f"idx: {idx}")
    print(f"address: {address}")
    print(f"*args: {args}")
    print(f"**kwargs: {kwargs}")
```

```python
personal_info6(1, '서울', 30,40,50, 이름 = '홍길동', 성별="남자")
```

* 1에서 10까지 수 중에서 3의 배수만 문자열로 바꾸기

```python
def tostr(x):
    result = []
    for n in x:
        if n % 3 == 0:
            result.append(str(n))
        else:
            result.append(n)
    return result
```

```python
tostr2 = lambda x: str(x) if x % 3 == 0 else x
```

```python
tostr2 = lambda x: str(x) if x % 3 == 0 else x

a = list(map(tostr2, list(range(1,11))))

print(type(a[0]))
a
```

# 예외처리 try exception
* 예외: 프로그램 실행 중에 발생할 수 있는 오류
* 3/0 : ZeroDivisionError -> 숫자를 0으로 나눌 수 없기 때문에 발생하는 에러
* try-except 구문을 사용하면 예외상황에서도 프로그램이 중단되지 않게 할 수 있음.

```python
for i in range(5):
    num = int(input("숫자를 입력해주세요"))
    try:
        x = 10/num
    except Exception as e:
        print(e)
        print("숫자 0으로는 나눗셈을 할 수 없습니다.")

    else:
        print(x)
```

```python
# 학생정보
student_scores = dict(alice=85, bob=92, mark=78)
```

```python
# 성적을 조회할 학생목록
names = ['alice','bob','mark','david']

# 반복문을 이용해 성적 조회
for name in names:
    try:
        score = student_scores[name]
        print(f"{name}의 점수는 {score}점 입니다.")
    except Exception as e:
        print(f"{name}의 점수는 없습니다.")
    finally:
        print(f"{name}의 점수 조회가 완료되었습니다.")
        print()
```

```python
import tensorflow as tf

# GPU 장치 이름을 가져옵니다
gpu_device_name = tf.test.gpu_device_name()

# GPU가 사용 가능한지 확인합니다
if gpu_device_name:
    print('Default GPU Device: {}'.format(gpu_device_name))
    print(tf.reduce_sum(tf.random.normal([1000, 1000])))
else:
    print("Please install GPU version of TF")
```

```python
from konlpy.tag import Mecab
mecab = Mecab()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.text import Tokenizer, text_to_word_sequence
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Input, Flatten, Embedding
from tensorflow.keras.utils import to_categorical
import re
import tensorflow as tf
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.metrics import classification_report
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, GlobalMaxPooling1D
from attention import Attention
from tensorflow.keras.layers import GlobalAveragePooling1D, LayerNormalization, MultiHeadAttention, Add
from tensorflow.keras.models import Model
```

```python
test_data = pd.read_csv("../05machine_learning/data/bank_app_reviews_test.csv")
test_data
```

```python
def clean_text(text):
    cleaned = re.sub(r'[^가-힣a-zA-Z0-9\s]','',text)  # 한글 영문 숫자 추출
    cleaned = re.sub(r'\s+',' ', cleaned) # 연속된 공백을 하나의 공백으로 줄임
    return cleaned.strip()
```

```python
test_data['사용자리뷰'] = test_data['사용자리뷰'].apply(clean_text)
test_data
```

```python
test_data['is_good'] = test_data['평점'].apply(lambda x: 1 if x >= 4 else 0)
test_data['is_good']
```

```python
tokenized_docs = test_data['사용자리뷰'].apply(mecab.morphs)
tokenized_docs[2]
```

```python
token = Tokenizer(lower=False)
token.fit_on_texts(tokenized_docs)
print(len(token.word_index))
```

```python
x = token.texts_to_sequences(tokenized_docs)
print(x[0])
```

```python
max_length = max([len(i) for i in x]) + 1
```

```python
X_padded = pad_sequences(x, maxlen=max_length, padding = 'post')
print(X_padded[0])
```

```python
X_padded.shape
```

```python
y = test_data['is_good']
y.value_counts()
```

```python
import joblib
```

```python
birnn_best_model = load_model("./model/bank_app_review_birnn.keras")
result = birnn_best_model.evaluate(X_padded, y)
print("loss ", result[0], "accuracy ", result[1])
```

```python
lstm_best_model = load_model("./model/bank_app_review_lstm_cnn.keras")
result = lstm_best_model.evaluate(X_padded, y)
print("loss ", result[0], "accuracy ", result[1])
```

```python
attn_best_model = load_model("./model/bank_app_review_attn_model.keras")
result = attn_best_model.evaluate(X_padded, y)
print("loss ", result[0], "accuracy ", result[1])
```

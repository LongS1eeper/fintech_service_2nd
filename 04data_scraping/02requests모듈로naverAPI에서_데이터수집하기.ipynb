# 2. 네이버 API로 검색한 결과 최대한 가져오기
* 1회 네이버 서버에 요청
* 응답에서 total을 찾아서 display로 나누어 총 반복할 페이지 산출
* 반복문 작성 및 종료 조건 설정
* 1페이지 결과를 데이터 프레임으로 만들고 리스트에 append
* 반복이 끝난 후 리스트에 append 된 데이터 프레임을 1개로 합침

```python
def unpack(data):
    book_dic = {}
    for item in data['items']:
        for key, value in item.items():
            book_dic.setdefault(key,[]).append(value)
    return book_dic

def unpack2(data, dic):
    for item in data['items']:
        for key, value in item.items():
            dic.setdefault(key,[]).append(value)
    return dic
```

### for 문 이용한 풀이

```python
keyword = input('제목을 입력하세요: ')

url = "https://openapi.naver.com/v1/search/book.json"
payload = {'query': keyword, 'display': 100, 'start':1, 'sort': 'sim'}  

headers = {'X-Naver-Client-Id' : "KWqT_uEHMFwBxZHzqvXj", 'X-Naver-Client-Secret' : 'UaIJQBMlNd'}

r = requests.get(url, params=payload, headers=headers)
data = r.json()     
book_dic = unpack(data)

page = data['total'] // 100 + 1

for i in range(101,1000,100):
    if data['total'] > page*100:
        break
    payload = {'query': keyword, 'display': 100, 'start':i, 'sort': 'sim'}  
    r = requests.get(url, params=payload, headers=headers)
    data = r.json()   
    book_dic = unpack2(data, book_dic)
    

book_df = pd.DataFrame(book_dic)    
book_df
```

# 블로그, 책, 뉴스 동시에 수집하기

```python
keyword = input('제목을 입력하세요: ')

book_dic = {}

for category in ['blog', 'book', 'news']:
    
    url = f"https://openapi.naver.com/v1/search/{category}.json"
    i = 1

    while (i <= 1000):
        payload = {'query': keyword, 'display': 100, 'start':i, 'sort': 'sim'}  
        headers = {'X-Naver-Client-Id' : "KWqT_uEHMFwBxZHzqvXj", 'X-Naver-Client-Secret' : 'UaIJQBMlNd'}
        r = requests.get(url, params=payload, headers=headers)

        data = r.json()
        for item in data['items']:
            for key, value in item.items():
                book_dic.setdefault(key,[]).append(value)

        if i >= (data['total']):
            break
        else:
            if i == 901:
                i += 99
            else:
                i += 100
            
# pd.concat() 합치는 함수

book_df = pd.DataFrame(book_dic)    
book_df
```

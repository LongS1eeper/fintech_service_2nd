```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup as bs

i = 1
j = 1

name_list = []
author_list = []
publisher_list = []
publishdate_list = []
cost_list = []
reviewscore_list = []
reviewnum_list = []

for item in soup.select('a.gd_name'):  # .gd_name도 가능
    # result.setdefault("도서명", []).append(item) -> 이런 식으로 바로 딕셔너리 형성도 가능
    name_list.append(item.text)

for item in soup.select('div.info_row.info_pubGrp'):
    author_list.append(item.select_one('span.authPub.info_auth a').text)

for item in soup.select('span.authPub.info_pub a'):
    publisher_list.append(item.text)

for item in soup.select('span.authPub.info_date'):
    publishdate_list.append(item.text)
    
for item in soup.select('strong.txt_num'):
    cost_list.append(item.text)
    
for item in soup.select('span.rating_grade em.yes_b'):
    reviewscore_list.append(item.text)
    if i == 10 or i == 19:
        reviewscore_list.append('미발매')
    i += 1
    
for item in soup.select('em.txC_blue'):
    reviewnum_list.append(item.text)
    if j == 10 or j == 19:
        reviewnum_list.append('미발매')
    j += 1
```

```python
total = [name_list, author_list, publisher_list, publishdate_list, cost_list, reviewscore_list, reviewnum_list]
column = ['도서명', '저자', '출판사', '출간일', '가격', '회원리뷰 점수', '회원리뷰 건수']
```

```python
for lists in total:
    print(len(lists))
```

```python
result_dic = {}

for col, value in zip(column, total):
    result_dic[col] = value

result_df = pd.DataFrame(result_dic)
result_df
```

```python
result = {}
column = ['book_title', 'author', 'publisher', 'date_pub', 'price', 'rating', 'n_reviews']

for i in range(1,10):
    url = 'https://www.yes24.com/product/category/bestseller'
    payload = dict(categoryNumber='001', pageNumber=i, pageSize=120)
    r = requests.get(url, params=payload)
    soup = bs(r.content, 'lxml')
    time.sleep(5)

    book_list = soup.select("ul#yesBestList > li")

    for book in book_list:
        book_title = book.select_one(".gd_name").text              # 책제목
        author = book.select_one(".info_row.info_pubGrp a").text   # 저자
        publisher = book.select_one(".authPub.info_pub a").text    # 출판사
        date_pub = book.select_one(".authPub.info_date").text      # 출간일
        price = int(book.select_one(".info_row.info_price em.yes_b").text.replace(',',''))    # 가격
        if (book.select_one(".rating_grade em.yes_b")):            # 평점
            rating = float(book.select_one(".rating_grade em.yes_b").text)
        else:
            rating = 0
        if (book.select_one(".info_row.info_rating em.txC_blue")): # 리뷰 수
            n_reviews = book.select_one(".info_row.info_rating em.txC_blue").text   
        else:
            n_reviews = 0

        values = [book_title, author, publisher, date_pub, price, rating, n_reviews]

        for col, val in zip(column, values):
            result.setdefault(col, []).append(val)

    
result_df = pd.DataFrame(result)
result_df
```

# 저자, 역자, 글그림, 편저 등 분류하기

```python
author =""
photo = ""
trans = ""

for item in text_clean(book_list[28].select_one(".authPub.info_auth").text).split('/'):
    if '저' == item[-1]:
        author = item
    elif '사진' == item[-2:]:
        photo = item
    elif '역' == item[-1]:
        trans = item
    print(f"author {author} photo {photo} trans {trans}")
```

```python
def text_clean(text):
    return text.replace("\n",' ').replace('\r',' ').strip
```

```python
import os
import gradio as gr
from google import genai

# 1) API 클라이언트 초기화
apikey = os.getenv("GEMINI_API_KEY")
client = genai.Client(api_key=apikey)

# 2) 사용자 입력 → Gemini 응답 함수
def respond(user_message, history):
    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=user_message
    )
    bot_message = response.text
    history = history + [(user_message, bot_message)]
    # (입력창 초기화, chat 히스토리, 상태 업데이트)
    return "", history, history

# 3) Gradio UI 정의
with gr.Blocks() as demo:
    gr.Markdown("## Google Gemini 챗봇")
    chatbot = gr.Chatbot(label="대화 내역")
    state   = gr.State([])  # 히스토리 저장용
    msg     = gr.Textbox(
        placeholder="메시지를 입력하세요...", 
        label="입력"
    )
    clear   = gr.Button("초기화")

    # Enter → respond 호출
    msg.submit(respond, [msg, state], [msg, chatbot, state])
    # 초기화 버튼 → 대화 히스토리 비우기
    clear.click(lambda: ([], []), None, [chatbot, state])

    # 포트포워딩 없이 로컬 7860 포트로 실행
    demo.launch()
```

- 위에서 볼 수 있듯이 링크를 타고 들어가서 기사 요약해서 만들어오기

```python
import math
import requests
import pandas as pd
from bs4 import BeautifulSoup
import torch
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
import gradio as gr

# ── 1) 뉴스 페치 & 전처리 ─────────────────────────────────
def fetch_news():
    url     = "https://openapi.naver.com/v1/search/news.json"
    params  = {'query': '핀테크', 'display': 100, 'start': 1, 'sort': 'sim'}
    headers = {
        'X-Naver-Client-Id':     "KWqT_uEHMFwBxZHzqvXj",
        'X-Naver-Client-Secret': 'UaIJQBMlNd'
    }
    r = requests.get(url, params=params, headers=headers)
    df = pd.DataFrame(r.json()['items'])
    df['title'] = df['title'].str.replace('&quot;','')\
                             .str.replace('<b>','')\
                             .str.replace('</b>','')
    return df

df       = fetch_news()
records  = df.to_dict('records')
PAGE_SIZE= 10
pages    = [records[i:i+PAGE_SIZE] for i in range(0, len(records), PAGE_SIZE)]
NUM_PAGES= len(pages)

# ── 2) 요약 파이프라인 로드 (FP16 + auto device map) ─────────
MODEL_NAME = "facebook/bart-large-cnn"
tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME)
model      = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)
summarizer = pipeline(
    "summarization",
    model=model,
    tokenizer=tokenizer,
    framework="pt"
)

# ── 3) 콜백 함수들 ───────────────────────────────────────
def get_page_titles(page_idx):
    return [item['title'] for item in pages[page_idx]]

def change_page(direction, state):
    page = state["page"]
    if direction=="next" and page<NUM_PAGES-1:
        page+=1
    elif direction=="prev" and page>0:
        page-=1
    state["page"] = page
    titles = get_page_titles(page)
    return (
        gr.Dropdown.update(choices=titles, value=titles[0] if titles else None),
        state,
        ""  # 요약 박스 초기화
    )

def summarize_article(selected_title, state):
    page = state["page"]
    link = next(
        (item['link'] for item in pages[page] if item['title']==selected_title),
        None
    )
    if not link:
        return "링크를 찾을 수 없습니다."
    resp = requests.get(link)
    soup = BeautifulSoup(resp.text, 'html.parser')
    paras = soup.find_all('p')
    full_text = " ".join(p.get_text() for p in paras)
    summary = summarizer(
        full_text,
        max_length=130, min_length=30, do_sample=False
    )[0]['summary_text']
    return summary[:300]

# ── 4) Gradio UI ────────────────────────────────────────
with gr.Blocks() as demo:
    gr.Markdown("## 핀테크 뉴스 300자 요약기")
    state = gr.State({"page": 0})

    with gr.Row():
        with gr.Column(scale=4):
            dropdown = gr.Dropdown(
                choices=get_page_titles(0),
                value=get_page_titles(0)[0],
                label="기사 제목"
            )
            with gr.Row():
                btn_prev = gr.Button("← 이전 페이지")
                btn_next = gr.Button("다음 페이지 →")
        with gr.Column(scale=8):
            with gr.Tabs():
                with gr.TabItem("300자 뉴스 요약"):
                    summary_box = gr.Textbox(lines=10, placeholder="여기에 요약이 표시됩니다.")

    btn_prev.click(
        change_page, inputs=[gr.State(state.value), state],
        outputs=[dropdown, state, summary_box]
    )
    btn_next.click(
        change_page, inputs=[gr.State(state.value), state],
        outputs=[dropdown, state, summary_box]
    )
    dropdown.change(
        summarize_article, inputs=[dropdown, state],
        outputs=summary_box
    )

demo.launch()
```

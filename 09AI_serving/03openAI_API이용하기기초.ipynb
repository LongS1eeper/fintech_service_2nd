# gradio를 이용한 openAI 챗봇

````python
import os
from openai import OpenAI
import gradio as gr

# ---------- OpenAI 클라이언트 ----------
client = OpenAI()  # OPENAI_API_KEY 환경 변수를 자동 인식

SYSTEM_PROMPT = "핀테크 전문가로서 대답해줘"

def chat_completion(user_input: str, chat_history: list):
    """
    Gradio가 호출하는 콜백 함수.
    chat_history: [(user_msg, assistant_msg), ...] 형태
    """
    # 과거 대화 기록 + 이번 사용자 메시지까지 묶어 전달
    messages = [{"role": "system", "content": SYSTEM_PROMPT}]
    for user_msg, assistant_msg in chat_history:
        messages.append({"role": "user", "content": user_msg})
        messages.append({"role": "assistant", "content": assistant_msg})
    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=0,
    )
    assistant_reply = response.choices[0].message.content
    return assistant_reply


# ---------- Gradio UI ----------
with gr.Blocks(title="핀테크 전문가 Chatbot") as demo:
    gr.Markdown("## 💸 핀테크 전문가에게 무엇이든 물어보세요!")
    chatbot = gr.Chatbot(height=450)
    user_box = gr.Textbox(
        placeholder="질문을 입력한 뒤 Enter",
        show_label=False,
    )
    clear_btn = gr.Button("대화 초기화")

    # 메시지 전송
    def user_submit(user_message, chat_history):
        if user_message.strip() == "":
            return gr.update(value=""), chat_history  # 빈 입력 방지
        assistant_message = chat_completion(user_message, chat_history)
        chat_history.append((user_message, assistant_message))
        return gr.update(value=""), chat_history

    user_box.submit(user_submit, inputs=[user_box, chatbot], outputs=[user_box, chatbot])
    clear_btn.click(lambda: None, None, chatbot, queue=False)

demo.launch()
````

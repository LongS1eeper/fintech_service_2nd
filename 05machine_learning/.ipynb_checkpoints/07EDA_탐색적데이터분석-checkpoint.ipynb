* 나이를 구간화해서 생존율과의 관계 파악
    * 10, 20, 30, 40, 50
    * 00 - 10 under10s
    * 10 - 19 10s
    - 20 - 29 20s

```python
def age_cat(x):
    if x < 10:
        return "under10s"
    elif 10 <= x < 20:
        return "10s"
    elif 20 <= x < 30:
        return "20s"
    elif 30 <= x < 40:
        return "30s"
    elif 40 <= x < 50:
        return "40s"
    elif 50 <= x < 60:
        return "50s"
    elif 60 <= x < 70:
        return "60s"
    elif 70 <= x < 80:
        return "70s"
    elif 80 <= x < 90:
        return "80s"
```

```python
# 수치형으로 번경
data['Age_cat'] = data['Age'].apply(age_cat)
data
```

# 문자로 되어있는 범주형 자료를 숫자로 변환하기
* 조건문을 이용해서 숫자로 변환하기
* label encoding을 이용해서 변환하기 - 나중에 숫자에서 문자로 복원 가능
* one-hot encoding unique value의 개수만큼 새로운 컬럼을 만들어서 0/1 구분 벡터
* male 1/0 female 1/0

* 반복문과 조건문을 이용해서 바꾸기

```python
result = []
for i in df['Sex']:
    if i == 'male':
        result.append(0)
    else:
        result.append(1)
df['Sex2'] = result
```

* .str.replace를 통해 바꾸기

```python
df['Sex3'] = df['Sex'].str.replace('female','1').str.replace('male','0')
df['Sex3'] = df['Sex3'].astype(int)
```

* 람다함수를 통해 바꾸기

```python
df['Sex'].apply(lambda x: 0 if x == 'male' else 1)
```

```python
df
```

* 결측값 제거

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.info()
```

```python
df = df.drop(['PassengerId', 'Name', 'Cabin'], axis=1)
```

```python
df = df.drop(['Ticket'], axis=1)
```

```python
df['Family'] = df['SibSp'] + df['Parch']
```

```python
df = df.drop(['SibSp', 'Parch'], axis = 1)
```

* one-hot encoding
* value의 개수만큼 새로운 컬럼(변수)을 만들어서 0/1 표시
* 자료의 서열이 사라짐
* 더미변수 만들기, get_dummies()
* value가 너무 많으면 변수의 수가 너무 많이 증가 문제 발생
* 변수가 너무 많이 늘어나면 **다중공선성**
* get_dummies(df,drop_first=True)

```python
df2 = pd.get_dummies(df, columns = ['Sex', 'Family', 'Embarked'], drop_first=True)
```

```python
X = df2.drop(['Survived', 'Fare', 'Sex2','Sex3'], axis=1)
```

```python
y = df2['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=10)
```

```python
print(len(X_train), len(X_test))
print(len(y_train), len(y_test))
X_train.head()
```

```python
dtc = DecisionTreeClassifier(random_state=10)
dtc.fit(X_train, y_train)
pred = dtc.predict(X_test)
pd.DataFrame({
    "y_test": y_test.reset_index(drop=True),
    "pred": pred
}).head(60)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
dtc = DecisionTreeClassifier(random_state=10)
dtc.fit(X_train, y_train)
pred = dtc.predict(X_test)
print(accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
rtc = RandomForestClassifier(random_state=10)
rtc.fit(X_train, y_train)
pred = rtc.predict(X_test)
print(accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
lr = LogisticRegression(random_state=10)
lr.fit(X_train, y_train)
pred = lr.predict(X_test)
print(accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

# 상관관계 분석 .corr()
* 변수들 간의 상관관계를 분석
* 1이면 강한 양의 상관관계 정비례
* -1이면 강한 음의 상관관계 반비례
* 0이면 관계가 없음
* 0.5, -0.5면 고민
* 0.7, -0.7이상이면 두 변수 중 한 개만 선택
* method = 'pearson' : 숫자형 데이터들 간의 상관관계 분석
* method = 'spearman' : 범주형 데이터가 섞여 있는 데이터셋의 경우 스피어만으로 상관관계 분석

```python
df2.corr(method='spearman')
```

* 상관관계를 보여주는 그래프: heatmap

```python
import seaborn as sns
```

```python
plt.figure(figsize=(20,20))
sns.heatmap(
    df2.corr(method='spearman'),
    annot=True,          # annot : 히트박스 속 숫자
    linewidth=0.5        #linewidth : 박스 사이 선
)    
```

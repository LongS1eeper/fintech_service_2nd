# 변수를 자동 생성하면서 긍정부정 데이터프레임 만들기
* globals()[변수명]

```python
bank_en_list = ['hana', 'shinhan', 'kb',  'toss', 'woori', 'b_salad']
```

```python
# 1) 결과를 모아둘 리스트 초기화
pos_neg_list = []

# 2) 은행 한글명과 영문명을 동시에 순회
for bank, bank_eng in zip(train_df['은행명'].unique(), bank_en_list):
    
    # 3) 해당 은행의 긍정 리뷰(+1)와 부정 리뷰(0)를 필터링
    pos = train_df[(train_df['은행명'] == bank) & (train_df['긍정'] == 1)]
    neg = train_df[(train_df['은행명'] == bank) & (train_df['긍정'] == 0)]
    
    # 4) 전역 변수 이름을 f"{영문명}_pos"/f"{영문명}_neg"로 자동 생성하여 할당
    globals()[f'{bank_eng}_pos'] = pos
    globals()[f'{bank_eng}_neg'] = neg
    
    # 5) 생성된 변수명을 문자열로 리스트에 추가 (나중에 일괄 참조용)
    pos_neg_list.append(f'{bank_eng}_pos')
    pos_neg_list.append(f'{bank_eng}_neg')
    
    # 6) 막대 그래프로 긍정 vs 부정 리뷰 수 시각화
    plt.bar(
        ['긍정', '부정'],                 # x축 레이블
        [pos.shape[0], neg.shape[0]],     # y축 값: 리뷰 개수
        color=['skyblue', 'salmon']       # 색상 지정
    )
    plt.title(f'{bank} 긍정부정 분포')      # 그래프 제목
    plt.ylabel('리뷰 수')                 # y축 이름
    # y축 범위 설정: 최대 리뷰 수 + 100 여유 공간
    plt.ylim(0, max(pos.shape[0], neg.shape[0]) + 100)
    plt.show()
```

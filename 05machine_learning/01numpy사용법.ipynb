## 차원
* 0차원 = 점. 스칼라 (a = 10)
* 1차원 = 선 --------- 백터, 배열, list, tuple, [1,2,3,4]
    * 벡터는 방향을 가진 데이터
* 2차원 = 면, 가로, 세로, DB 테이블, Dataframe, matrix
* 3차원 = 입체, 가로, 세로, 높이, 정육면체
* 4차원 = 3차원 요소가 2개 이상
* 5차원 = 4차원 요소가 2개 이상

```python
# 스칼라 연산
a = 10
b = 5
scalar = a + b
print(scalar)
```

```python
# 1차원 데이터 벡터의 연산
b1 = [1,2,3,4,5]
# 리스트 요소에 10을 곱한 후 다시 리스트로 만드세요.
result = []
for n in b1 :
    result.append(n*10)
print(result)
```

```python
result2 = [num*10 for num in b1]
result2
```

```python
# 2차원 데이터 계산
second = np.array([[1,2,3],[1,2,3]])
print(second, second.ndim, second.shape)
```

```python
result = []
for num in second:
    mid_result = []
    for nu in num:
        mid_result.append(num*10)
    result.append(mid_result)
result = np.array(result)
result
```

```python
# 3차원 데이터의 요소에 10 곱하기
third = np.array([[[1,2,3],[1,2,3]],[[1,2,3],[1,2,3]]])
```

```python
result = []
for num1 in third:
    num1_result = []
    for num2 in num1:
        num2_result = []
        for num3 in num2:
            num2_result.append(num3*10)
        num1_result.append(num2_result)
    result.append(num1_result)
result = np.array(result)
result
```

```python
# 4차원 데이터 (아래에서 numpy가 효율적인 이유를 볼 수 있음)
fourth = np.array([[[[1,2,3],[1,2,3]],[[1,2,3],[1,2,3]]], [[[1,2,3],[1,2,3]],[[1,2,3],[1,2,3]]]])
```

```python
result = []
for num1 in fourth:
    num1_result = []
    for num2 in num1:
        num2_result = []
        for num3 in num2:
            num3_result = []
            for num4 in num3:
                num3_result.append(num4*10)
            num2_result.append(num3_result)
        num1_result.append(num2_result)
    result.append(num1_result)
result = np.array(result)
result
```

```python
fourth*10
```

```python
# axis = -1 가장 안쪽 차원 1차원 배열을 정렬 (행 정렬)
# axis = 1 2차원 배열에서 각 열을 정렬 (위아래만 비교)
# axis = 0 다른 2차원 배열과 값 비교
# np.sort() 함수의 axis 기본값은 -1
arr = np.array([[[5,2,8],[1,6,7]],[[4,9,3],[0,2,1]]])
print(arr)
print('-1')
print(np.sort(arr, axis=-1))
print('0')
print(np.sort(arr, axis=0))
print('1')
print(np.sort(arr, axis=1))
```

```python
# argsort() 자료 중 값이 작은 것에서 큰 순으로 원래 인덱스 출력
g = np.array([5,0.5,-7,42,10,0])
print(g)
```

```python
s = g.argsort()
s
```

```python
g[s][::-1]
```

# 배열 합치기

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])
print(a,a.shape)
print(b, b.shape)
```

* np.concatenate((a,b))
* 합치는 기본 방향은 axis=0 행이 늘어남, 아래로 붙음
* axis = 1 열이 늘어남, 옆으로 붙음

```python
np.concatenate((a,b))
```

```python
np.concatenate((a,b), axis=1)
```

### 합치는 방향의 모양이 일치하지 않으면 합치기 X

```python
c = np.array([5,6])
print(c,c.ndim,c.shape)
```

```python
np.concatenate((a,c))
```

```python
d = np.array([[7,8,9]])
print(d, d.ndim, d.shape)
```

```python
f = np.array([[1,2],[3,4],[5,6]])
print(f, f.ndim, f.shape)
```

```python
np.concatenate((e,f), axis=1)
```

```python
# 전치 행렬: 행과 열을 바꿈
print(f)
print(f.shape)
print(f.T)
print(f.T.shape)
```

```python
np.concatenate((e,f.T), axis=1)    # 이제 붙을 수 있음
```

```python
np.concatenate((e,f.T))    # 이제 붙을 수 있음
```
